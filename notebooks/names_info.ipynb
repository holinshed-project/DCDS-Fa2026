The following notebook contains basic information about names extracted from:
- `holinshed_names.csv`: All extracted names.
- `holinshed_names_unique.csv`: Unique extracted names.

```python
import pandas as pd

df_holinshed_names = pd.read_csv("../name_extraction/entity_names/holinshed_names.csv")
df_holinshed_names_unique = pd.read_csv("../name_extraction/entity_names/holinshed_names_unique.csv")

df_holinshed_names.info()
df_holinshed_names_unique.info()
```

```python
n_unique_holinshed_names = df_holinshed_names['name'].nunique()
print(f"holinshed_names unique count: {n_unique_holinshed_names}")

# Should print the same number of unique names as in df_holinshed_names_unique
n_unique_holinshed_names_unique = df_holinshed_names_unique['name'].nunique()
print(f"holinshed_names_unique count: {n_unique_holinshed_names_unique}")
```

```python
# Find names in holinshed_names that are not in holinshed_names_unique
df_names_not_in_unique = df_holinshed_names[~df_holinshed_names['name'].isin(df_holinshed_names_unique['name'])]
print(f"{df_names_not_in_unique.shape[0]} names are in holinshed_names but not in holinshed_names_unique")
df_names_not_in_unique.head(10)
```
